```python
class Solution(object):
    def search(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: int
        """
        left = 0
        right = len(nums) -  1
        # if len(nums) == 1:
        #     if nums[0] == target:
        #         return 0
        #     else:
        #         return -1

        # The `while left <= right:` loop in the provided code is a binary search algorithm. 
        # It continues to iterate as long as the `left` pointer is less than or equal to the `right` pointer. 
        # This loop is used to search for the target element within the given sorted `nums` 
        # list by adjusting the `left` and `right` pointers based on the comparison of the target value with the middle element of the current range.
        while left <= right:
            # means:

            # “Keep searching while there are still numbers left to check.”

            # left is the start of the search area
            # right is the end of the search area

            # As long as left has not passed right, there is still a possible place where the target could be.
            mid = (left + right) // 2
            if nums[mid] == target:
                return mid
            if target > nums[mid]:
                left =  mid + 1
            else:
                right = mid - 1
        return -1
```

```python
def isPalindrome(string):
    if string == string[::-1]:
        return True
    return False


def isPalindrome2(string):
    reversed = '' 
    for i in string:
        reversed = i + reversed
    if string == reversed:
        return True
    return False

print(isPalindrome("racecar"))
print(isPalindrome("hello"))

print(isPalindrome2("racecar"))
print(isPalindrome2("hello"))
```
